# Resonance vs. Integrator Dynamics in the AMAT Model

This notebook investigates and validates the characteristics of Post-Synaptic Currents (PSCs) in resonators versus integrators using the AMAT model. It reproduces the findings, neuronal characteristics, and visual data presented in the Appendix of Yamauchi et al., 2011.

### Difference between integrator vs. resonator PSC behaviour? 
- Integrator Behaviour (All-Pass): Acts like a leaky bucket. It accumulates (sums up) all incoming inputs over time, regardless of their timing. It fires a spike when the total input volume exceeds the threshold. It acts as a low-pass filter, favoring high-frequency bursts or closely packed inputs.
- Resonator Behaviour (Band-Pass): Acts like a swing. It is highly sensitive to the specific **timing** and **frequency** of inputs. It prefers a specific inter-pulse interval ($\Delta$). If a second pulse arrives at the peak of the internal oscillation caused by the first pulse, the system "resonates" and triggers a spike. If the timing is off, the inputs cancel out.

These characteristics which we aim to simulate are displayed in this figure below. 


<img src="images/integrator_resonator.png" width="1050" height="600">

## Creating and Running a Simple Integrator/Resonator AMAT Simulation

The purpose of this experiment is to reproduce the **resonator versus integrator behaviour** described by Yamauchi et al. (2011). Rather than investigating repetitive spiking, this experiment examines how the neuron's **dynamic firing threshold** evolves in response to a pair of brief current pulses separated by different time intervals.

### Dynamic threshold in the AMAT model

Unlike a standard leaky integrate-and-fire neuron, the AMAT model uses a **dynamic firing threshold**,

$\theta(t)=\omega+\theta_{\alpha_1}(t)+\theta_{\alpha_2}(t)+\theta_V(t)$,


where

* $\omega$ is the resting (baseline) firing threshold,
* $\theta_{\alpha_1}(t)$ is the fast spike-history threshold component,
* $\theta_{\alpha_2}(t)$ is the slow spike-history threshold component,
* $\theta_V(t)$ is the voltage-dependent threshold component introduced by Yamauchi et al. (2011).

A spike is emitted whenever

$V_m(t)\geq\theta(t)$

where $V_m(t)$ is the membrane potential.

Throughout this notebook, we therefore monitor

$V_m(t)-\theta(t),$

which measures how close the neuron is to firing.

* If $V_m-\theta<0$, the neuron remains below threshold.
* If $V_m-\theta=0$, the neuron reaches threshold.
* If $V_m-\theta>0$, the neuron emits a spike.

---

### Resonator versus integrator behaviour

The AMAT model can behave either as

* an **integrator**, where inputs simply accumulate until threshold is reached, or
* a **resonator**, where the neuron's excitability depends on the timing between successive inputs.

This behaviour is controlled by two dimensionless parameters:

$\gamma=\frac{\tau_\beta}{\tau_m}$

and

$B=\beta\tau_m,$

where

* $\tau_m$ is the membrane time constant,
* $\tau_\beta$ is the time constant governing the voltage-dependent threshold,
* $\beta$ determines the strength of the voltage-dependent feedback.

For implementation, these relationships are rearranged to

$\tau_\beta=\gamma\tau_m$

$\beta=\frac{B}{\tau_m}$

By varying $\gamma$ and $B$, the neuron transitions between integrator and resonator regimes.

---

### Experimental protocol

To analyse this behaviour, Yamauchi et al. apply **two identical current pulses** separated by an inter-pulse interval, $\Delta$.

In the analytical derivation, each pulse is represented as a Dirac delta current,

$I(t)=\delta(t)$

where $\delta(t)$ represents an infinitesimally short current impulse. This idealised input allows the membrane and threshold dynamics to be solved analytically.

The simulation keeps the following quantities fixed:

* current pulse amplitude,
* current pulse duration,
* $\beta$,
* $\gamma$,

The objective is to determine whether a second pulse arriving after a particular delay moves the membrane potential above the dynamic threshold.

---

### Why are the spike-history terms inactive?

During the resonance experiment, the neuron is intentionally kept **below threshold** for most of the simulation. The reason for doing this is to isolate the effect of the **voltage-dependent threshold mechanism**. If the neuron were to spike immediately after the first pulse, the spike-history terms

$
\theta_{\alpha_1}(t)
\quad\text{and}\quad
\theta_{\alpha_2}(t)
$

would both increase, making it impossible to determine whether the observed resonance originated from the new voltage-dependent mechanism or simply from conventional spike-triggered adaptation.

Consequently, before the neuron emits its first spike,

* $\theta_{\alpha_1}(t)=0$,
* $\theta_{\alpha_2}(t)=0$,

and the resonance behaviour is determined almost entirely by the interaction between the membrane potential,

$
V_m(t),
$

and the voltage-dependent threshold component,

$
\theta_V(t).
$

Only if the second current pulse arrives within a favourable time window does

$
V_m(t)-\theta(t)
$
cross zero, producing an output spike. This selective response to the timing between two identical inputs is the defining characteristic of the resonator regime.

```python
def make_pulse_current(pulse_times, pulse_width, pulse_amplitude):
    amplitude_times = []
    amplitude_values = []

    for t0 in pulse_times:
        amplitude_times.extend([t0, t0 + pulse_width])
        amplitude_values.extend([pulse_amplitude, 0.0])

    return nest.Create(
        "step_current_generator",
        params={
            "amplitude_times": amplitude_times,
            "amplitude_values": amplitude_values,
        },
    )


def run_amat_current(
    model_name,
    pulse_times,
    pulse_amplitude=500.0,
    pulse_width=1.0,
    simtime=500.0,
    gamma=0.5,
    B=0.5,
    omega=5.0,
    alpha_1=10.0,
    alpha_2=0.0,
    verbose=True,
):
    nest.ResetKernel()
    nest.set_verbosity("M_ERROR")

    try:
        nest.Install(module_name)
    except Exception:
        pass

    tau_m = nest.GetDefaults(model_name)["tau_m"]
    tau_v = gamma * tau_m
    beta = B / tau_m

    parameters = {
        "omega": omega,
        "alpha_1": alpha_1,
        "alpha_2": alpha_2,
        "beta": beta,
        "tau_v": tau_v,
    }

    neuron = nest.Create(model_name, params=parameters)

    current = make_pulse_current(
        pulse_times=pulse_times,
        pulse_width=pulse_width,
        pulse_amplitude=pulse_amplitude,
    )

    nest.Connect(current, neuron)

    sr = nest.Create("spike_recorder")
    nest.Connect(neuron, sr)

    if model_name == "amat2_psc_exp":
        recordables = ["V_m", "V_th", "V_th_v", "I_syn_ex"]
    else:
        recordables = [
            "V_m",
            "V_th_alpha_1",
            "V_th_alpha_2",
            "V_th_v",
            "V_th_v_aux",
        ]

    mm = nest.Create(
        "multimeter",
        params={
            "interval": 0.1,
            "record_from": recordables,
        },
    )

    nest.Connect(mm, neuron)
    nest.Simulate(simtime)

    mm_events = mm.get("events")
    sr_events = sr.get("events")

    full_parameters = {
        **parameters,
        "gamma": gamma,
        "B": B,
        "tau_m": tau_m,
    }

    if verbose:
        print(f"\n-Debugging Statements for: {model_name}")
        print("Parameters:", full_parameters)
        print("Spikes logged:", sr_events["times"])
        print("Vm maximum recorded value:", np.max(mm_events["V_m"]))
        print("Vm minimum recorded value:", np.min(mm_events["V_m"]))

    return mm_events, sr_events, full_parameters


pulse_times = [200.0, 220.0, 400.0, 460.0]

print("\nRunning NESTML AMAT...")
NESTML_AMAT_df, NESTML_AMAT_spikes, NESTML_AMAT_parameters = run_amat_current(
    neuron_model_name,
    pulse_times=pulse_times,
    pulse_amplitude=500.0,
    pulse_width=1.0,
    simtime=500.0,
    gamma=0.5,
    B=0.5,
    omega=5.0,
    alpha_1=10.0,
    alpha_2=0.0,
)

print("\nRunning NEST AMAT...")
NEST_AMAT_df, NEST_AMAT_spikes, NEST_AMAT_parameters = run_amat_current(
    "amat2_psc_exp",
    pulse_times=pulse_times,
    pulse_amplitude=500.0,
    pulse_width=1.0,
    simtime=500.0,
    gamma=0.5,
    B=0.5,
    omega=5.0,
    alpha_1=10.0,
    alpha_2=0.0,
)
```

# Delta-Sweep Experiment 

For every Δ:
- inject pulse at 200 ms
- inject second pulse at 200 + Δ
- record whether the neuron spikes

### What are we expecting? 

```python
def run_delta(
    model_name,
    delta,
    pulse_amplitude=625.0,
    pulse_width=0.5,
    gamma=0.5,
    B=0.5,
    omega=5.0,
    alpha_1=10.0,
    alpha_2=2.0):
    
    pulse_times_delta = [200.0, 200.0 + delta]

    data, spikes, params = run_amat_current(
        model_name,
        pulse_times=pulse_times_delta,
        pulse_amplitude=pulse_amplitude,
        pulse_width=pulse_width,
        simtime=350.0,
        gamma=gamma,
        B=B,
        omega=omega,
        alpha_1=alpha_1,
        alpha_2=alpha_2,
        verbose=False,
    )

    t = data["times"]
    V_m = data["V_m"]
    E_L = nest.GetDefaults(model_name).get("E_L", -70.0)

    theta = (
        E_L
        + params["omega"]
        + data["V_th_alpha_1"]
        + data["V_th_alpha_2"]
        + data["V_th_v"]
    )

    distance = V_m - theta

    second_pulse_time = 200.0 + delta
    response_window = (t >= second_pulse_time) & (t <= second_pulse_time + 60.0)

    max_distance_after_second = np.max(distance[response_window])

    spike_times = spikes["times"]
    spike_after_second = np.any(
        (spike_times >= second_pulse_time)
        & (spike_times <= second_pulse_time + 60.0)
    )

    return {
        "delta": delta,
        "max_distance_after_second": max_distance_after_second,
        "spike_after_second": spike_after_second,
        "data": data,
        "spikes": spikes,
        "distance": distance,
        "theta": theta,
        "params": params,
    }


deltas = np.arange(2.0, 150.0, 1.0)

results = [
    run_delta(
        neuron_model_name,
        delta=d,
        pulse_amplitude=625.0,
        pulse_width=0.5,
        gamma=0.5,
        B=0.5,
        omega=5.0,
        alpha_1=10.0,
        alpha_2=2.0,
    )
    for d in deltas
]

max_distance = np.array([r["max_distance_after_second"] for r in results])
spike_after_second = np.array([r["spike_after_second"] for r in results])


fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(deltas, max_distance, marker="o", markersize=3)
ax.axhline(0, linestyle="--")
ax.set_xlabel("Inter-pulse interval Δ (ms)")
ax.set_ylabel(r"max post-second-pulse $V_m-\theta$ (mV)")
ax.set_title("NESTML AMAT delta tuning curve")
plt.tight_layout()
plt.show()


fig, ax = plt.subplots(figsize=(8, 2.5))
ax.plot(deltas, spike_after_second.astype(int), marker="o", markersize=3)
ax.set_xlabel("Inter-pulse interval Δ (ms)")
ax.set_ylabel("Spike?")
ax.set_yticks([0, 1])
ax.set_title("NESTML Spike occurrence after second pulse")
plt.tight_layout()
plt.show()
```
