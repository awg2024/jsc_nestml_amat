# Visualising Non-Scaling Benchmark

Order 2500 varying num compute nodes. 

```python
import pandas as pd
import matplotlib.pyplot as plt

# load your benchmarking spreadsheet
data_path = "data/amat_bench_nonscale.csv"
df = pd.read_csv(data_path)

# set up the plotting window
plt.figure(figsize=(8, 5))

# plot both model types against the nodes column
plt.plot(df['num_cores'], df['nest_build'], marker='o', linewidth=2, label='NEST')
plt.plot(df['num_cores'], df['nestml_build'], marker='s', linewidth=2, label='NESTML')

plt.xlabel('Number of Compute Nodes', fontsize=11)
plt.ylabel('Simulation Time (seconds)', fontsize=11)

# Start Y-axis at zero to accurately show scaling improvements
plt.ylim(bottom=0)

# Add layout enhancers
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()


plt.savefig('non_scale_bench.png', dpi=100)
print("Finished")
```
